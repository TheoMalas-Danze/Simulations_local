```julia
const ∞ = Inf
ħ=1
e=1
β = ∞ # Case zero temperature T=0, β → ∞
ϕ_quantum=1/2

#### Caracterization of the circuit #####

ω_a = 3.8533*2*pi
ω_b = 2*ω_a# 7.0515*2*pi


φ_a = 0.11
Z_a = 2*ϕ_quantum^2*φ_a^2
C_a_inv = Z_a*ω_a
E_Ca = C_a_inv/2

φ_b = 0.204
Z_b = 2*ϕ_quantum^2*φ_b^2
C_b_inv = Z_b*ω_b
E_Cb = C_b_inv/2

E_J = 37.12*2*pi*1.6


##Drive


ω_d=ω_b
ω_p=2ω_a-ω_d

ϵ_0 = 0.3#0.61
g2 = besselj(1,ϵ_0)*E_J*φ_a^2*φ_b # (Leon's paper)
ϵ_d=10.12*g2

buffer_phase = 0.0
pump_phase = 0.0;

##Env

κ_b=1/10.4; # 1/ns from Leon's paper
κ_2 = 4*g2^2/κ_b
κ_a = 0;#3e-3*κ_2 #from Leon's paper
```

$$ H = \omega_a a^*a + \omega_b b^*b -2E_J\sin(\varepsilon_0)\sin\biggl(\frac{a+a^*+2\Delta_a}{\sqrt 2} \frac{ \sqrt Z_a}{ϕ_q} + \frac{b+b^*+2\Delta_b}{\sqrt 2} \frac{ \sqrt Z_b}{ϕ_q}\biggl)$$
$$ H = \omega_a \biggl(\frac{p_a^2}{2} + \frac{x_a^2}{2}\biggl) + \omega_b \biggl(\frac{p_b^2}{2} + \frac{x_b^2}{2}\biggl) -2E_J\sin(\varepsilon_0)\sin\biggl((x_a+\sqrt 2\Delta_a) \frac{ \sqrt Z_a}{ϕ_q} + (x_b+\sqrt 2\Delta_b) \frac{ \sqrt Z_b}{ϕ_q}\biggl)$$
$$ H = T + U $$
$$ T = \omega_a\frac{p_a^2}{2} + \omega_b \frac{p_b^2}{2} $$
$$ U = \omega_a \frac{x_a^2}{2} + \omega_b \frac{x_b^2}{2} -2E_J\sin(\varepsilon_0)\sin\biggl(x_a \frac{ \sqrt Z_a}{ϕ_q} + x_b \frac{ \sqrt Z_b}{ϕ_q} + \varphi_d\biggl)$$

$$ U = \omega_a \frac{x_a^2}{2} + \omega_b \frac{x_b^2}{2} + \varepsilon_d \cos(\omega_b t)x_b -2E_J\sin(\varepsilon_0)\sin\biggl(x_a \frac{ \sqrt Z_a}{ϕ_q} + x_b \frac{ \sqrt Z_b}{ϕ_q} + \varphi_d\biggl)$$
$\varphi_d = -\frac{2E_J\sin(\varepsilon_0)}{\omega_a \phi_q^2}(Z_a+Z_b/2)$
$$ U = \omega_a \frac{x_a^2}{2} + \omega_b \frac{x_b^2}{2} -2E_J\sin(\varepsilon_0)\sin\biggl(x_a \frac{ \sqrt Z_a}{ϕ_q} + x_b \frac{ \sqrt Z_b}{ϕ_q} + \varphi_d + \xi\cos(\omega_b t)\biggl)$$
$\xi = \frac{2\varepsilon_d(\omega_b-i\kappa_b)}{\omega_b^2-(\omega_b-i\kappa_b)^2}=\frac{2\varepsilon_d(\omega_b-i\kappa_b)}{\kappa_b^2+2i\omega_b\kappa_b}$

```julia
x_a = -3 : 0.01 : 3
x_b = -3: 0.01 : 3

xa = repeat(reshape(x_a, :, 1), 1, length(x_b))
xb = repeat(reshape(x_b, 1, :), length(x_a), 1)

sin_pump =  - 2*E_J*sin(ϵ_0)
Δ_a = sqrt(Z_a)/(sqrt(2)*ϕ_quantum*ω_a)*sin_pump
Δ_b = sqrt(Z_b)/(sqrt(2)*ϕ_quantum*ω_b)*sin_pump  

# Define potential U(φa, φb)
φ_d = 0#sin_pump *(Z_a + Z_b/2)/ϕ_quantum^2 / ω_a
U = @. ω_a*xa^2/2 + ω_b*xb^2/2 - 2E_J*sin(ϵ_0)sin(xa/ϕ_quantum + xb/ϕ_quantum + φ_d);
```
