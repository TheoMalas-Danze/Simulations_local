```julia
x=-2pi:0.01:2pi
y=Taylor_sin.(x,5)
```

```julia
plot(x,sin.(x), label="sin")
for i in 3:2:13
    plot!(x,Taylor_sin.(x,i), label=string(i))
end
vline!([0.5],linestyle=:dash, color=:indianred1, label="")
vline!([-0.5],linestyle=:dash, color=:indianred1, label="")
annotate!(0.7,-2, Plots.text("convergence", 8, :indianred1, rotation=90))
vline!([2],linestyle=:dash, color=:red, label="")
vline!([-2],linestyle=:dash, color=:red, label="")
annotate!(2.2,-2, Plots.text("second lobe", 8, :red, rotation=90))
vline!([4],linestyle=:dash, color=:darkred, label="")
vline!([-4],linestyle=:dash, color=:darkred, label="")
annotate!(4.2,-2, Plots.text("first lobe", 8, :darkred, rotation=90))
plot!(xlims=(-2pi,2pi), ylims=(-5,5))
```
