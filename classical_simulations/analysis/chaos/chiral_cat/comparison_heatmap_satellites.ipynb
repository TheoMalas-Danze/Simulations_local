```julia
const ∞ = Inf
ħ=1
e=1
β = ∞ # Case zero temperature T=0, β → ∞
ϕ_quantum=1/2

#### Caracterization of the circuit #####

ω_a = 25.338776456203686
ω_b_0 = 2*ω_a

Lamb_shift = 0.05 #Lamb shift in GHz


φ_a = 0.11#*(rand(1)[1]/2+0.75)
Z_a = 2*ϕ_quantum^2*φ_a^2
C_a_inv = Z_a*ω_a
E_Ca = C_a_inv/2

φ_b = 0.204#*(rand(1)[1]/2+0.75)
Z_b = 2*ϕ_quantum^2*φ_b^2
C_b_inv = Z_b*ω_b_0
E_Cb = C_b_inv/2

E_J = 37.12*2*pi*1.5#*(rand(1)[1]/2+0.75)


##Drive


ω_d=ω_b_0#+Lamb_shift*2*pi # Drive frequency
ω_p=2ω_a-ω_d

ϵ_0 = 0.4581818181818182
g2 = besselj(1,ϵ_0)*E_J*φ_a^2*φ_b # (Leon's paper)
ϵ_d=10.124276*g2

buffer_phase = 0.0
pump_phase = 0.0;

##Env

κ_b=1/10.4#*(rand(1)[1]/2+0.75); # 1/ns from Leon's paper
κ_2 = 4*g2^2/κ_b
κ_a = 0#3e-3*κ_2 #from Leon's paper
```

```julia
ϵ_p_list = 0.2:.025:1.1
ϵ_d_g2_ratio_list = 4:.2:13.5;
```

```julia
using JSON
satellite_position_dict = JSON.parsefile("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/satellites_all_12_02_2026_Hill.json")
Re_a_dict = Dict{Float64, Float64}()
Im_a_dict = Dict{Float64, Float64}()
Re_b_dict = Dict{Float64, Float64}()
Im_b_dict = Dict{Float64, Float64}()
for ϵ_p in ϵ_p_list
    Re_a_dict[ϵ_p] = satellite_position_dict[string(ϵ_p)]["a_R_real"][75]/φ_a
    Im_a_dict[ϵ_p] = satellite_position_dict[string(ϵ_p)]["a_R_imag"][75]/φ_a
    Re_b_dict[ϵ_p] = satellite_position_dict[string(ϵ_p)]["b_R_real"][75]/φ_b
    Im_b_dict[ϵ_p] = satellite_position_dict[string(ϵ_p)]["b_R_imag"][75]/φ_b
end
```

```julia
t_max = 50pi / (ω_a)#1 / (50 * ω_a)#130/κ_b#
t_min = 0.0
t_min_save = t_max - 30/κ_b
tspan = (0.0, t_max) # Time span for the simulation
dt = 1 / (50 * ω_a) # Time step for the simulation

dt_save=2pi/(30*sqrt(2)*ω_a)
t_save = max(t_min_save,0):dt_save:t_max


b_0=0.0
environment = MeanField(κ_a, κ_b)
system = SystemATS_ladder(ω_a, ω_b_0, E_J, E_Ca, E_Cb, 0.0, b_0)

environment_prepared = prepare(system,environment);
t_list = collect(range(max(t_min_save,0),step = dt_save,stop = t_max));
```

```julia
Final_all_a_bis = Dict{Float64,Dict{Float64, Vector{ComplexF32}}}()
Final_all_b_bis = Dict{Float64,Dict{Float64, Vector{ComplexF32}}}()
Threads.@threads for ϵ_p in ϵ_p_list
    g2 = besselj(1,ϵ_p)*E_J*φ_a^2*φ_b # (Leon's paper)
    Final_mode_a = Dict{Float64, Vector{ComplexF32}}()
    Final_mode_b = Dict{Float64, Vector{ComplexF32}}()
    for ratio in ϵ_d_g2_ratio_list
        ϵ_d=ratio*g2
        drive = DriveTwoModesBasics(ϵ_d, ω_d, buffer_phase, ϵ_p, ω_p, pump_phase)
        drive_prepared = prepare(drive)

        re_a_i = Re_a_dict[ϵ_p]    
        im_a_i = Im_a_dict[ϵ_p]
        re_b_i = Re_b_dict[ϵ_p]    
        im_b_i = Im_b_dict[ϵ_p]
            
            
        a_0_i = (re_a_i + 1im*im_a_i)*exp(-1im*0.05)
        b_0_i = (re_b_i + 1im*im_b_i)*exp(-1im*0.1)

        system = SystemATS_ladder(ω_a, ω_b_0, E_J, E_Ca, E_Cb, a_0_i, b_0_i)
        system_prepared = prepare(system)

        u_0 = get_init(system_prepared, environment_prepared)
        p = SimulationParams(system_prepared, drive_prepared, environment_prepared)

        a, b, t_list = default_solver(u_0, (t_min, t_max), p, dt, t_save = t_save, equation = default_equation!)

        a_rot = a.*exp.(1im*ω_a*t_list)
        b_rot = b.*exp.(1im*ω_b_0*t_list)

        Final_mode_a[ratio] = a_rot
        Final_mode_b[ratio] = b_rot
    end
    Final_all_a_bis[ϵ_p] = Final_mode_a
    Final_all_b_bis[ϵ_p] = Final_mode_b
end
```

```julia
for ϵ_p in ϵ_p_list
    g2 = besselj(1,ϵ_p)*E_J*φ_a^2*φ_b
    Final_mode = Final_all_a_bis[ϵ_p]
    p0 = plot()

    # Overlay scatter points
    for ratio in ϵ_d_g2_ratio_list
        scatter!(p0,
                    real.(Final_mode[ratio]),
                    imag.(Final_mode[ratio]),
                    ms=.5, label="", color=:black)
    end

    # Add labels and title
    plot!(p0,
          xlabel=L"\Re(a)",
          ylabel=L"\Im(a)",
          title=L"g_2/2\pi="*string(round(g2/2pi*1000,digits=3))*" MHz",
          top_margin=5mm)

    display(p0)
end
```

```julia
t_max = 10/κ_b
re_a_0, im_a_0 = 0, 0
ϕ_a_0, q_a_0=sqrt(2*Z_a)*re_a_0, sqrt(2/Z_a)*im_a_0
re_b_0, im_b_0 = 0, 0
ϕ_b_0, q_b_0=sqrt(2*Z_b)*re_b_0, sqrt(2/Z_b)*im_b_0

system = SystemATS(ω_a, ω_b_0, E_J, E_Ca, E_Cb, ϕ_a_0, q_a_0, ϕ_b_0, q_b_0)
environment = MeanField(κ_a, κ_b) #OneChain(n, chain_params...)

system_prepared = prepare(system)
environment_prepared = prepare(system,environment)

dt = get_dt(system, environment)
u_dot_0, u_0 = get_init(system_prepared, environment_prepared)
u_0_new = vcat(u_0,u_dot_0);
```

```julia
Lyap_coef_chaos = Array{Float64}(undef, length(ϵ_d_g2_ratio_list), length(ϵ_p_list))
drive = DriveTwoModesJA(0.0, ω_d, buffer_phase, 0.0, ω_p, pump_phase)
drive_prepared = prepare(drive)
t_w = 5/κ_b
for i in ProgressBar(1:length(ϵ_p_list))
    ϵ_p = ϵ_p_list[i]
    drive.ϵ_0 = ϵ_p
    g2_max_i = besselj(1,ϵ_p)*E_J*φ_a^2*φ_b # (Leon's paper)

    re_a_i = Re_a_dict[ϵ_p]    
    im_a_i = Im_a_dict[ϵ_p]
    re_b_i = Re_b_dict[ϵ_p]    
    im_b_i = Im_b_dict[ϵ_p]
        
        
    a_0_i = (re_a_i + 1im*im_a_i)*exp(-1im*0.05)
    b_0_i = (re_b_i + 1im*im_b_i)*exp(-1im*0.1)

    re_a_0, im_a_0 = real(a_0_i), imag(a_0_i)
    ϕ_a_0, q_a_0=sqrt(2*Z_a)*re_a_0, sqrt(2/Z_a)*im_a_0
    re_b_0, im_b_0 = real(b_0_i), imag(b_0_i)
    ϕ_b_0, q_b_0=sqrt(2*Z_b)*re_b_0, sqrt(2/Z_b)*im_b_0

    system = SystemATS(ω_a, ω_b_0, E_J, E_Ca, E_Cb, ϕ_a_0, q_a_0, ϕ_b_0, q_b_0)
    environment = MeanField(κ_a, κ_b) #OneChain(n, chain_params...)

    system_prepared = prepare(system)
    environment_prepared = prepare(system,environment)

    dt = get_dt(system, environment)
    u_dot_0, u_0 = get_init(system_prepared, environment_prepared)
    u_0_new = vcat(u_0,u_dot_0);
    for j in 1:length(ϵ_d_g2_ratio_list)
        ϵ_d_j = ϵ_d_g2_ratio_list[j]*g2_max_i
        drive.ϵ_d = ϵ_d_j
        drive_prepared.drive = drive
        p = SimulationParams(system_prepared, drive_prepared, environment_prepared)

        problem = CoupledODEs(equation_1ODE!, u_0_new, p)

        Lyap_coef_chaos[j,i] = lyapunov(problem, t_max, Δt=dt; d0 = 1e-7, d0_upper = 1e-4, Ttr = t_w)
    end
end
Lyap_coef_chaos_sign = map(x -> x>0 ? 1 : -1, Lyap_coef_chaos);
```
