```python
import pandas as pd

df = pd.read_csv("../output/torus_25_11_2025.csv")
x = df["x"].values
y = df["y"].values
z = df["z"].values
t = df["t"].values
```

```python
from ipywidgets import interact
import matplotlib.pyplot as plt
import numpy as np

%matplotlib widget

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Scatter returns the mappable object
sc = ax.scatter(x, y, z, c=t/30, cmap='viridis', s=1)

ax.set_xlabel(r'$Re(a)\cos(\theta)$')
ax.set_ylabel(r'$Re(a)\sin(\theta)$')
ax.set_zlabel(r'$Im(a)$')

# Attach colorbar to the scatter
fig.colorbar(sc, ax=ax, label=r'$\kappa_bt$')

plt.show()
```

```python
t_sorted = np.sort(t)
x_time_sorted = np.array([x_i for _,x_i in sorted(zip(t,x))])
y_time_sorted = np.array([y_i for _,y_i in sorted(zip(t,y))])
z_time_sorted = np.array([z_i for _,z_i in sorted(zip(t,z))])
```

```python
start = 0
stop = 100
step = 1

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(x[::10], y[::10], z[::10], c="black", s=.5)
# Scatter returns the mappable object
sc = ax.scatter(x_time_sorted[start:stop:step], y_time_sorted[start:stop:step], z_time_sorted[start:stop:step], c=t_sorted[start:stop:step]/30, cmap='RdYlBu', s=15)

ax.set_xlabel(r'$Re(a)\cos(\theta)$')
ax.set_ylabel(r'$Re(a)\sin(\theta)$')
ax.set_zlabel(r'$Im(a)$')

# Attach colorbar to the scatter
fig.colorbar(sc, ax=ax, label=r'$\kappa_bt$')

plt.show()
```

```python
import json

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/data_logistic.json", "r") as f:
    data = json.load(f)

x_all = data["x_all"]
y_all = data["y_all"]
z_all = data["z_all"]
```

```python
import random
idx = random.sample(range(0, len(x_all)), 50000)
x_all_bis = [x_all[i] for i in idx]
y_all_bis = [y_all[i] for i in idx]
z_all_bis = [z_all[i] for i in idx]
```

```python
from ipywidgets import interact
import matplotlib.pyplot as plt
import numpy as np

%matplotlib widget

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Scatter returns the mappable object
sc = ax.scatter(x_all_bis, y_all_bis, z_all_bis, s=15, color =(51/255, 101/255, 138/255))

ax.set_xlabel("\n"+r'$Re(a)$', fontsize=30)
ax.set_ylabel("\n"+r'$Im(a)$', fontsize=30)
ax.set_zlabel("\n"+r'$\kappa_a$', fontsize=30)

ax.tick_params(axis='both', which='major', labelsize=20)
ax.tick_params(axis='z', which='major', labelsize=20)

plt.tight_layout()
plt.show()
```
