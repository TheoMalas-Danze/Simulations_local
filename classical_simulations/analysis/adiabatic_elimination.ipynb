# In the case of a sine expanded up to 5th order with only resonant terms

$$H(t)=\delta_a a^\dagger a + \delta_b b^\dagger b + \frac{\epsilon_d}{2}(b^\dagger + b)+(g+\alpha a^\dagger a + \beta b^\dagger b)(a^2b^\dagger + H.c.)$$

Displacement with $\gamma=-\frac{\epsilon_d/2}{\delta_b+\kappa_b/2i}$

$$U^\dag \mathcal{L} U = -i[\delta_a a^\dagger a + \delta_b b^\dagger b +(g+\alpha a^\dagger a + \beta (b^\dagger+\gamma^*) (b+\gamma))(a^2(b^\dagger+\gamma^*) + H.c.), \cdot]+\kappa_b\mathcal{D}[b]$$

```python
import sympy as sp

# Declare symbols (mean-field variables, treated as complex numbers)
a, a_dag = sp.symbols("a a_dag", complex=True)
b, b_dag = sp.symbols("b b_dag", complex=True)

g, alpha, beta = sp.symbols("g alpha beta", real=True)
delta_a, delta_b = sp.symbols("delta_a delta_b", real=True)
kappa_b = sp.symbols("kappa_b", real=True)
gamma = sp.symbols("gamma", complex=True)

# Hermitian conjugates of gamma (constant)
gamma_conj = sp.conjugate(gamma)

# Define Hamiltonian (mean-field)
H_eff = (
    delta_a * a_dag * a
    + delta_b * b_dag * b
    + (g + alpha*a_dag*a + beta*(b_dag + gamma_conj)*(b + gamma))
        * (a**2*(b_dag + gamma_conj) + a_dag**2*(b + gamma))
)

# Equations of motion
a_dot = -sp.I * sp.diff(H_eff, a_dag)
b_dot = -sp.I * sp.diff(H_eff, b_dag) - kappa_b * b / 2
```

```python
b_dot
```

```python
poly = sp.Poly(b_dot, b, b_dag)

A = poly.coeffs()[ poly.monoms().index((0,0)) ] if (0,0) in poly.monoms() else 0
B = poly.coeffs()[ poly.monoms().index((1,0)) ] if (1,0) in poly.monoms() else 0
C = poly.coeffs()[ poly.monoms().index((0,1)) ] if (0,1) in poly.monoms() else 0
D = poly.coeffs()[ poly.monoms().index((2,0)) ] if (2,0) in poly.monoms() else 0
E = poly.coeffs()[ poly.monoms().index((1,1)) ] if (1,1) in poly.monoms() else 0
```

```python
x, y = sp.symbols("x y", real=True)
A_re, A_im, B_re, B_im, C_re, C_im, D_re, D_im, E_re, E_im = sp.symbols("A_re, A_im, B_re, B_im, C_re, C_im, D_re, D_im, E_re, E_im", real=True)

subs_b = {
    b:  x + sp.I*y,
    b_dag: x - sp.I*y,
    A : A_re + sp.I*A_im,
    B : B_re + sp.I*B_im,
    C : C_re + sp.I*C_im,
    D : D_re + sp.I*D_im,
    E : E_re + sp.I*E_im,
}
b_eq = A + B*b + C*b_dag + D*b**2 + E*b*b_dag
eq_real = sp.simplify(sp.re(b_eq.subs(subs_b)))
eq_imag = sp.simplify(sp.im(b_eq.subs(subs_b)))
```

```python
sol_eq_real = sp.solve(eq_real, x)
```

```python
x_0 = sol_eq_real[0]
eq_imag = sp.expand(eq_imag.subs(x, x_0))
```

```python
import sympy as sp

# Variables
x, y = sp.symbols('x y', real=True)
A1, B1x, C1x, B1y, C1y = sp.symbols("A1, B1x, C1x, B1y, C1y", real=True)
A2, B2x, C2x, B2y, C2y = sp.symbols("A2, B2x, C2x, B2y, C2y", real=True)

# Example polynomial form
# Replace A1, B1x, C1x, B1y, C1y with your actual parameters
F1 = A1 + B1x*x + C1x*x**2 + B1y*y + C1y*y**2
F2 = A2 + B2x*x + C2x*x**2 + B2y*y + C2y*y**2

# Compute Gröbner basis (lexicographic order eliminates y or x)
G = sp.groebner([F1, F2], x, y, order='lex')

# Extract basis
g1, g2 = list(G)
# Solve the triangular system:
sol_y = sp.solve(g1, y)   # g1 is univariate in 
solutions = []
```

```python
yy = sol_y[0]
eq_x = g2.subs(y, yy)
```

```python
eq_x
```

```python

for yy in sol_y:
    
    xs = sp.solve(eq_x, x)
    for xx in xs:
        solutions.append({x: xx, y: yy})

print(solutions)
```
