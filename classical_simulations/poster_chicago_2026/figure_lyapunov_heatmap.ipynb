```python
with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/g2_list_07_04_2026_Hill_unstable.json", "r") as f:
    g2_unstable = np.array(json.load(f))

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/data_lyapunov_07_04_2026.json", "r") as f:
    data_lyapunov = np.array(json.load(f))

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/g2_list_07_04_2026.json", "r") as f:
    g2_list = np.array(json.load(f))/2/np.pi*1e3

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/alpha_sq_list_07_04_2026.json", "r") as f:
    alpha_sq = json.load(f)

alpha_sq_Hill = np.linspace(5,13.5,40)/2
```

```python
from matplotlib.colors import ListedColormap, BoundaryNorm

X, Y = np.meshgrid(g2_list, alpha_sq, indexing='ij')

# Discrete colormap: blue = regular, red = chaos
cmap = ListedColormap([(13/256, 76/256, 116/256), (185/256, 21/256, 21/256)])

# Define boundaries between values
norm = BoundaryNorm([-1.5, 0, 1.5], cmap.N)

plt.figure()
pcm = plt.pcolormesh(X, Y, np.array(data_lyapunov), cmap=cmap, norm=norm)

plt.xlabel(r"$g_2$ (MHz)")
plt.ylabel(r"$\varepsilon_d/2g_2$")

plt.plot(g2_unstable, alpha_sq_Hill[::-1], color="k", label="Unstability of the satellites")

# Colorbar with custom labels
cbar = plt.colorbar(pcm, ticks=[-1, 1])
cbar.ax.set_yticklabels(["Regular", "Chaotic"])


plt.legend(loc="upper center", bbox_to_anchor=(0.5, 1.2), frameon=False)
#plt.tight_layout()
plt.savefig("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/figures/poster_chicago/lyapunov_heatmap.svg")
plt.show()
```
