```python
import numpy as np
import json
import sys
import matplotlib.pyplot as plt
sys.path.insert(0, "/Users/tmalas/Desktop/Dev/Simulations_local/Code_python")
import Plot_setting
from Plot_setting import *
```

```python
# 12/10/2026: maps and Hill line with g2 = E_J sin(eps_p) phi_a^2 phi_b, eps_d/g2 in 2:0.1:13.5
with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/g2_list_12_10_2026_Hill_unstable.json", "r") as f:
    g2_unstable = np.array(json.load(f), dtype=float)          # MHz, along the ascending eps_d/g2 below

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/eps_d_ratio_list_12_10_2026_Hill_unstable.json", "r") as f:
    alpha_sq_Hill = np.array(json.load(f))/2

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/data_lyapunov_12_10_2026.json", "r") as f:
    data_lyapunov = np.array(json.load(f))

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/g2_list_12_10_2026.json", "r") as f:
    g2_list = np.array(json.load(f))/2/np.pi*1e3

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/alpha_sq_list_12_10_2026.json", "r") as f:
    alpha_sq = np.array(json.load(f))
```

```python
# chaos onset: for each g2 (row of data_lyapunov), the smallest eps_d/2g2 with a positive exponent (NaN if none)
alpha_sq_chaos = np.array([alpha_sq[np.where(row == 1)[0][0]] if np.any(row == 1) else np.nan for row in data_lyapunov])

plt.plot(g2_list, alpha_sq_chaos)
```

```python
from matplotlib.colors import ListedColormap, BoundaryNorm

X, Y = np.meshgrid(g2_list, alpha_sq, indexing='ij')

# Discrete colormap: blue = regular, red = chaos
cmap = ListedColormap([(13/256, 76/256, 116/256), (185/256, 21/256, 21/256)])

# Define boundaries between values
norm = BoundaryNorm([-1.5, 0, 1.5], cmap.N)

plt.figure()
pcm = plt.pcolormesh(X, Y, np.array(data_lyapunov), cmap=cmap, norm=norm)

plt.ylabel(r"$\varepsilon_d/2g_2$")
plt.xlabel(r"$g_2/2\pi$ (MHz)")

plt.plot(g2_unstable, alpha_sq_Hill, color="k", label="Instability of the satellites")

plt.plot(g2_list, alpha_sq_chaos, linewidth = 1, color=(242/256, 199/256, 67/256))

# points a, b, c of the phase-space panels (phase_space_trajectories.ipynb): (g2/2pi in MHz, eps_d/2g2)
for (g2_pt, alpha_sq_pt), name in zip([(60, 3), (88, 3), (105, 3)], "abc"):
    plt.plot(g2_pt, alpha_sq_pt, "o", ms=5, mfc="white", mec="k", mew=1)
    plt.annotate(name, (g2_pt, alpha_sq_pt), xytext=(0, 5), textcoords="offset points", ha="center", va="bottom",
                 color="white", fontweight="bold")

# Colorbar with custom labels
cbar = plt.colorbar(pcm, ticks=[-1, 1])
cbar.ax.set_yticklabels(["Regular", "Chaotic"])


plt.legend(loc="upper center", bbox_to_anchor=(0.5, 1.2), frameon=False)
#plt.tight_layout()
plt.savefig("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/figures/article/lyapunov_heatmap_origin_start_12_10_2026.pdf", bbox_inches="tight")
plt.show()
```
