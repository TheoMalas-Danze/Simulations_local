```python
import sys
sys.path.insert(0, "/Users/tmalas/Desktop/Dev/Simulations_local/Code_python")
import Plot_setting
import matplotlib.pyplot as plt
import numpy as np
import json
from scipy.special import jv
from Plot_setting import Perso_colormap
```

```python
INF = np.inf

hbar = 1
e = 1
beta = INF  # Zero-temperature case: T = 0, beta -> inf
phi_quantum = 1 / 2

phi_a = 0.11
phi_b = 0.204
E_J = 37 * 2 * np.pi * 1.5

epsilon_p_list = np.concatenate((np.arange(0.175, 0.6, 0.07),np.arange(0.6, 1.1, 0.03)))
g2_list = jv(1, epsilon_p_list) * E_J * phi_a**2 * phi_b

ratio = np.arange(1.5, 10.5, 1)
alpha_car = ratio / 2

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/otoc_example_22_04_2026.json", "r") as f:
    otoc_example = np.array(json.load(f)).T

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/otoc_example_stable_22_04_2026.json", "r") as f:
    otoc_example_stable = np.array(json.load(f)).T

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/lyap_array_22_04_2026_bis.json", "r") as f:
    data_lyap = np.array(json.load(f)).T

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/R_sq_array_22_04_2026_bis.json", "r") as f:
    data_R_sq = np.array(json.load(f)).T

with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/end_otoc_array_22_04_2026_bis.json", "r") as f:
    data_end_OTOC = np.array(json.load(f)).T
```
