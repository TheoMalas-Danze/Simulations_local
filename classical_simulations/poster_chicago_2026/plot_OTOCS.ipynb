## OTOC rate vs classical Lyapunov exponent, both started on a cat lobe (20/10/2026, larger cats added 21/10/2026)

Quantum: `lyap_best_init_coherent_15_10_2026.json` ($\varepsilon_d/2g_2 = 1$–$4.5$, best truncation per point, up to $n_a = 67$) and `otoc_largecat_na65_nb33_init_coherent_21_10_2026.json` ($\varepsilon_d/2g_2 = 5$–$7$). Classical: `data_lyapunov_cat_start_12_10_2026.json`.
Self-contained: only needs `phi_a`, `phi_b` and the imports of the first two cells.

```python
# Quantum OTOC rate (coherent start |i alpha>|0>) against the classical Lyapunov map started on the
# same cat lobe, a(0) = i|alpha|, b(0) = 0 (lyapunov_heatmap_cat_start.ipynb). Both on g2 = E_J sin(eps_p) phi_a^2 phi_b,
# eps_d = ratio g2, |alpha|^2 = ratio/2, so no relabelling (the g2_list of the cells above is still on J1).
# Differences that remain: point vs coherent state; classical exponent averaged over 30/kappa (code) after a
# 15/kappa transient, quantum rate fitted on kappa_b T = 0.225 (6 drive periods).
from matplotlib.colors import TwoSlopeNorm, ListedColormap

out_dir = "/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/analysis/output/chicago_poster_2026/"
E_J_sin = 37.12 * 2 * np.pi * 1.5
g2_MHz = lambda eps_p: np.sin(eps_p) * E_J_sin * phi_a**2 * phi_b / (2 * np.pi) * 1e3

with open(out_dir + "lyap_best_init_coherent_15_10_2026.json") as f:
    otoc_best = json.load(f)
eps_p_q = np.round(np.array(otoc_best["epsilon_0"]), 2)   # stored from float32
alpha_sq_q = np.array(otoc_best["ratio"]) / 2
lyap_q = np.array(otoc_best["lyap"], dtype=float)            # [eps_p][ratio], m/2 in units of kappa_b
# larger cats (21/10/2026): ratio 9..14 at n_a = 65, n_b = 33, one dict per point; ratio 9 repeats the last column above
with open(out_dir + "otoc_largecat_na65_nb33_init_coherent_21_10_2026.json") as f:
    otoc_large = json.load(f)
ratio_large = sorted({r["ratio"] for r in otoc_large} - set(otoc_best["ratio"]))
lyap_large = np.full((len(eps_p_q), len(ratio_large)), np.nan)
for r in otoc_large:
    if r["ratio"] in ratio_large and "error" not in r:
        lyap_large[np.argmin(abs(eps_p_q - r["epsilon_0"])), ratio_large.index(r["ratio"])] = r["lyap"]
alpha_sq_q = np.concatenate((alpha_sq_q, np.array(ratio_large) / 2))
lyap_q = np.hstack((lyap_q, lyap_large))
g2_q = g2_MHz(eps_p_q)

with open(out_dir + "data_lyapunov_cat_start_12_10_2026.json") as f:
    sign_c = np.array(json.load(f), dtype=float)             # [eps_p][ratio]
with open(out_dir + "g2_list_cat_start_12_10_2026.json") as f:
    g2_c = np.array(json.load(f)) / (2 * np.pi) * 1e3
with open(out_dir + "alpha_sq_list_cat_start_12_10_2026.json") as f:
    alpha_sq_c = np.array(json.load(f))

# classical sign on the quantum grid (both grids are on 0.01 in eps_p and integer ratios)
i_c = [int(np.argmin(abs(g2_c - g))) for g in g2_q]
j_c = [int(np.argmin(abs(alpha_sq_c - a))) for a in alpha_sq_q]
sign_c_on_q = sign_c[np.ix_(i_c, j_c)]
sign_q = np.sign(lyap_q)
print(f"same sign on {(sign_c_on_q == sign_q).sum()}/{sign_q.size} points; "
      f"quantum chaotic only: {((sign_q > 0) & (sign_c_on_q < 0)).sum()}, "
      f"classical chaotic only: {((sign_q < 0) & (sign_c_on_q > 0)).sum()}")

def onset(chaotic, g2):
    """Smallest g2 with a positive exponent (nan if none)."""
    k = np.flatnonzero(chaotic)
    return g2[k[0]] if k.size else np.nan

print("|alpha|^2   onset g2/2pi (MHz): quantum   classical (cat start)")
for j, a in enumerate(alpha_sq_q):
    print(f"{a:9.1f}   {onset(lyap_q[:, j] > 0, g2_q):29.1f}   {onset(sign_c[:, j_c[j]] > 0, g2_c):9.1f}")

y_lim = (alpha_sq_q[0] - 0.25, alpha_sq_q[-1] + 0.25)
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True, constrained_layout=True)

ax = axes[0]
c0 = ax.pcolormesh(g2_c, alpha_sq_c, sign_c.T, cmap=ListedColormap(["#2b6cb0", "#c53030"]),
                   shading="auto", vmin=-1, vmax=1)
ax.contour(g2_q, alpha_sq_q, lyap_q.T, levels=[0], colors="yellow", linewidths=2)
ax.plot([], [], color="yellow", lw=2, label="OTOC rate = 0")
ax.set_title("Classical, start on a cat lobe")
cbar0 = fig.colorbar(c0, ax=ax, ticks=[-0.5, 0.5])
cbar0.ax.set_yticklabels([r"$\lambda<0$", r"$\lambda>0$"])

ax = axes[1]
c1 = ax.pcolormesh(g2_q, alpha_sq_q, lyap_q.T, cmap="RdBu_r",
                   norm=TwoSlopeNorm(vcenter=0, vmin=lyap_q.min(), vmax=lyap_q.max()), shading="auto")
onset_c = np.array([onset(sign_c[:, j] > 0, g2_c) for j in range(len(alpha_sq_c))])
ax.plot(onset_c, alpha_sq_c, color="black", lw=1.5, label="classical onset")
ax.set_title(r"Quantum, start in $|i\alpha\rangle|0\rangle$")
fig.colorbar(c1, ax=ax).set_label(r"OTOC rate $/\kappa_b$")

for ax in axes:
    ax.set_xlim(g2_q[0], g2_q[-1])
    ax.set_ylim(*y_lim)
    ax.set_xlabel(r"$g_2/2\pi$ (MHz)")
    ax.legend(loc="upper left", fontsize=9)
axes[0].set_ylabel(r"$\varepsilon_d/2g_2$")
# plt.savefig("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/figures/article/otoc_vs_lyapunov_cat_start_21_10_2026.pdf")
plt.show()
```

```python
# Classical OTOC on the quantum window (classical_otoc_window.jl): C_cl = (d x_a^rot(t)/d x_a(0))^2 from the tangent dynamics
# of the map EOM, fitted as the quantum OTOC (rate = slope of log C vs kappa_b t / 2). "TWA" averages C_cl over the Wigner
# function of |i alpha>|0>; "point" starts at a(0) = i|alpha| as the cat-start map. Needs the previous cell (lyap_q, onset, ...).
with open(out_dir + "classical_otoc_window_21_10_2026.json") as f:
    cl = json.load(f)
assert np.allclose(np.array(cl["epsilon_0"]), eps_p_q) and np.allclose(np.array(cl["ratio"]) / 2, alpha_sq_q)
rate_twa = {w: np.array(r) for w, r in zip(cl["windows_twa"], cl["rate_twa"])}
rate_point = {w: np.array(r) for w, r in zip(cl["windows_point"], cl["rate_point"])}

panels = [(lyap_q, r"Quantum OTOC, 3 drive periods"),
          (rate_twa[3], "Classical TWA, 3 drive periods"),
          (rate_twa[30], "Classical TWA, 30 drive periods"),
          (rate_point[300], r"Classical, start at $i|\alpha|$, 300 drive periods")]

print("|alpha|^2   onset g2/2pi (MHz): quantum | TWA 3 | TWA 30 | point 3 | point 30 | point 300 | map (cat start)")
for j, a in enumerate(alpha_sq_q):
    cols = [lyap_q, rate_twa[3], rate_twa[30], rate_point[3], rate_point[30], rate_point[300]]
    print(f"{a:9.1f}   " + " | ".join(f"{onset(c[:, j] > 0, g2_q):6.1f}" for c in cols)
          + f" | {onset(sign_c[:, j_c[j]] > 0, g2_c):6.1f}")

fig, axes = plt.subplots(2, 2, figsize=(13, 9), sharex=True, sharey=True, constrained_layout=True)
for ax, (rate, title) in zip(axes.flat, panels):
    lim = np.nanmax(abs(rate))
    c = ax.pcolormesh(g2_q, alpha_sq_q, rate.T, cmap="RdBu_r", norm=TwoSlopeNorm(vcenter=0, vmin=-lim, vmax=lim),
                      shading="auto")
    fig.colorbar(c, ax=ax).set_label(r"rate $/\kappa_b$")
    ax.contour(g2_q, alpha_sq_q, lyap_q.T, levels=[0], colors="yellow", linewidths=2)
    ax.plot(onset_c, alpha_sq_c, color="black", lw=1.5)
    ax.set_title(title)
    ax.set_xlim(g2_q[0], g2_q[-1])
    ax.set_ylim(*y_lim)
axes[0, 0].plot([], [], color="yellow", lw=2, label="quantum OTOC rate = 0")
axes[0, 0].plot([], [], color="black", lw=1.5, label="classical map onset (cat start)")
axes[0, 0].legend(loc="upper left", fontsize=9)
for ax in axes[1]:
    ax.set_xlabel(r"$g_2/2\pi$ (MHz)")
for ax in axes[:, 0]:
    ax.set_ylabel(r"$\varepsilon_d/2g_2$")
# plt.savefig("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/figures/article/otoc_classical_window_21_10_2026.pdf")
plt.show()
```
