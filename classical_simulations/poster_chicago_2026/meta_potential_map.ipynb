```julia
const ∞ = Inf
ħ=1
e=1
β = ∞ # Case zero temperature T=0, β → ∞
ϕ_quantum=1/2
#### Caracterization of the circuit #####
ω_a = 25.338776456203686
ω_b = 2*ω_a

φ_a = 0.11
Z_a = 2*ϕ_quantum^2*φ_a^2
C_a_inv = Z_a*ω_a
E_Ca = C_a_inv/2

φ_b = 0.204
Z_b = 2*ϕ_quantum^2*φ_a^2
C_b_inv = Z_b*ω_b
E_Cb = C_b_inv/2

E_J = 37.12*2pi*1.5

#Environment
κ_b=1/10.4; # 1/ns from Leon's paper

ω_d=ω_b # Drive frequency
ω_p=2*ω_a-ω_d # Pump Frequency

ε_p = 0.2
g2 = besselj(1,ε_p)*E_J*φ_a^2*φ_b # (Leon's paper)
ε_d = 8.12*g2#0.75*g2_max

β_a_p, β_b_p = 2*E_J*sin(ε_p)*φ_a^2/(ω_a), 2*E_J*sin(ε_p)*φ_b^2/(ω_b)
ε_d_p = ε_d*φ_b/ω_b
κ_b_p = κ_b / ω_b

ℵ = β_a_p*β_b_p/(4*κ_b_p)
ℶ = β_a_p*ε_d_p/(2*κ_b_p)
```

```julia
x_all = -1:0.01:1
p_all = -3:0.01:3

U = [ℵ*φ_a^2*(x^2 + p^2)^2/4 + ℶ*(x^2 - p^2)/2 for p in p_all, x in x_all]

# coarser grid for quiver
xq = -3:3/3:3
pq = -.9:.5:.9

# U = ℵ/4*(x^2+p^2)^2 + ℶ/2*(x^2-p^2)
dUdx(x,p) = -ℵ*x*(x^2 + p^2)*φ_a^2 + ℶ*x
dUdp(x,p) = -ℵ*p*(x^2 + p^2)*φ_a^2 - ℶ*p

# negative gradient flow
vx = [dUdx(x,p) for p in pq, x in xq]
vp = [dUdp(x,p) for p in pq, x in xq];
```

```julia

heatmap(p_all, x_all, transpose(U)*1e3,
    colorbar_title="\n Meta-potential (MHz)",
    xlabel=L"$\mathrm{Re}(a)$",
    ylabel=L"$\mathrm{Im}(a)$",
    guidefontsize=15,
    tickfontsize=12,
    cmap = Perso_colormap
)

#  quiver!(
#      repeat(collect(xq)', length(pq), 1),
#      repeat(collect(pq), 1, length(xq)),
#      quiver=(vx*3, vp*3),
#      xlabel=L"$\Re(a)$",
#      ylabel=L"$\Im(a)$",
#      aspect_ratio=:equal,
#      legend=false,
#      color="white"
#  )

scatter!(
    [-sqrt(ℶ/ℵ)/φ_a, sqrt(ℶ/ℵ)/φ_a],
    [0.0, 0.0],
    marker=:x,
    ms=10,
    color=RGB(185/256, 21/256, 21/256),
    label="",
    msw=5,
)

#savefig("/Users/tmalas/Desktop/Presentations/Poster_Chicago_2026/meta_potential_flow_one_mode.svg")
```
