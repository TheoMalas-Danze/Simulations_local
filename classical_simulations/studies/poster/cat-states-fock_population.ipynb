```python
import dynamiqs as dq
import matplotlib.pyplot as plt
import sys
sys.path.append("/Users/tmalas/Desktop/Dev/Simulations_local/Code_python")
import Plot_setting
n = 20
psi_left = dq.coherent(n, -2)
psi_right = dq.coherent(n, 2)
psi_plus = psi_left+psi_right
psi_minus = psi_left-psi_right
```

```python
fig, ax = plt.subplots(1)
dq.plot.fock(psi_plus, ax=ax, ymax=None)
ax.set_ylabel(r"$\langle\hat n_a\rangle$", fontsize=25)
ax.tick_params(axis='both', labelsize=16)
ax.set_xlim(-1,13)
plt.tight_layout()
plt.savefig("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/figures/poster_chicago/fock_cat_plus.pdf")
plt.show()
```

```python
fig, ax = plt.subplots(1)
dq.plot.fock(psi_minus, ax=ax, ymax=None)
ax.set_ylabel(r"$\langle\hat n_a\rangle$", fontsize=25)
ax.tick_params(axis='both', labelsize=16)
ax.set_xlim(-1,13)
plt.tight_layout()
plt.savefig("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/figures/poster_chicago/fock_cat_minus.pdf")
plt.show()
```
