```python
import numpy as np
import json
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import sys
sys.path.append("/Users/tmalas/Desktop/Dev/Simulations_local/Code_python/")
import Plot_setting

# Example data
with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/data/limit_cycles/stable_trajectory_memory_re_09_04_2026.json", "r") as f:
    x_full = json.load(f)
with open("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/data/limit_cycles/stable_trajectory_memory_im_09_04_2026.json", "r") as f:
    y_full = json.load(f)
```

```python
y = np.array(x_full[2000:5000:3])*0.11
x = np.array(y_full[2000:5000:3])*0.11
```

```python
plt.plot(x,y)
plt.xlabel(r"$\Re(a)$")
plt.ylabel(r"$\Im(a)$")
```

```python
fig, ax = plt.subplots()
ax.set_xlim(1.1*np.min(x), 1.1*np.max(x))
ax.set_ylim(1.1*np.min(y), 1.1*np.max(y))
ax.set_xlabel(r"$\Re(a)$")
ax.set_ylabel(r"$\Im(a)$")
ax.set_aspect('equal')   # optional
ax.grid(True)

line, = ax.plot([], [], lw=2)
point, = ax.plot([], [], 'ro', ms=4)

def init():
    line.set_data([], [])
    point.set_data([], [])
    return line, point

def update(frame):
    line.set_data(x[:frame+1], y[:frame+1])
    point.set_data([x[frame]], [y[frame]])
    return line, point

ani = FuncAnimation(
    fig,
    update,
    frames=len(x),
    init_func=init,
    blit=True,
    interval=30
)

ani.save("/Users/tmalas/Desktop/Dev/Simulations_local/classical_simulations/figures/poster_chicago/stable_trajectory_memory_09_04_2026.mp4", writer="ffmpeg", fps=90, dpi=200)
```
