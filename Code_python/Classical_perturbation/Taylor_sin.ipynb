```python
from scipy.special import factorial
import sympy as sp
import numpy as np
def Taylor_sin(x,n):
    result = 0
    for i in range(n):
        result += ((-1)**i * x**(2*i + 1)) / factorial(2*i + 1)
    return result
phi_a = sp.symbols('phi_a')
phi_b = sp.symbols('phi_b')
a = sp.symbols('a')
a_st = sp.symbols('a^*')
b = sp.symbols('b')
b_st = sp.symbols('b^*')
Taylor_sin_expr = Taylor_sin(phi_a+phi_b, 3)
expr_3 = Taylor_sin_expr.replace(phi_a, (a+a_st)/np.sqrt(2)).replace(phi_b, (b+b_st)/np.sqrt(2))
expr_2 = Taylor_sin(phi_a+phi_b, 2).replace(phi_a, (a+a_st)/np.sqrt(2)).replace(phi_b, (b+b_st)/np.sqrt(2))
```

```python
terms = (expr_3-expr_2).expand().args
```
