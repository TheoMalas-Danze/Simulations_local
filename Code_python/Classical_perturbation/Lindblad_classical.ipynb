```python
import sympy as sp
import numpy as np
from IPython.display import Math

import sys
import os
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..')))
import Plot_setting
from Plot_setting import *
from Utils import * 
```

```python
hbar = sp.symbols('hbar')
x = sp.symbols('x',real=True)
p = sp.symbols('p', real=True)

C = sp.Function('C', complex=True)(x,p)
C_st = sp.Function('C^*', complex=True)(x,p)
rho = sp.Function('rho', real=True)(x,p)


def poisson_bracket(expr_1, expr_2, variables):
    result = 0
    for coupled_var in variables:
        var_1, var_2 = coupled_var
        result += (sp.diff(expr_1, var_1) * sp.diff(expr_2, var_2) - sp.diff(expr_2, var_1) * sp.diff(expr_1, var_2))*(-1j)
    return result


def star_product(f,g):
    return f*g + sp.I*hbar/2*poisson_bracket(f,g,[(x,p)])-hbar**2/8*(sp.diff(sp.diff(f,x),x)*sp.diff(sp.diff(g,p),p)\
            -2*sp.diff(sp.diff(f,x),p)*sp.diff(sp.diff(g,x),p)+sp.diff(sp.diff(f,p),p)*sp.diff(sp.diff(g,x),x))
```

```python
expr = star_product(star_product(C,rho),C_st)/(sp.I*hbar**2)-1/2*1/(sp.I*hbar)*(star_product(rho,star_product(C_st,C)/(sp.I*hbar))+ star_product(star_product(C_st,C)/(sp.I*hbar),rho))
```

```python
expr_simp = sp.simplify(sp.expand(expr))
```

```python
rho_0_j_re = (expr_simp*hbar**2).coeff(hbar,2)
```

```python
rho_0_j_re_conj = rho_0_j_re.conjugate()

replacement_dict = {
    sp.conjugate(C): C_st,
    sp.conjugate(C_st): C,
    sp.conjugate(rho): rho,
    sp.conjugate(sp.diff(C, x)): sp.diff(C_st, x),
    sp.conjugate(sp.diff(C, p)): sp.diff(C_st, p),
    sp.conjugate(sp.diff(rho, x)): sp.diff(rho, x),
    sp.conjugate(sp.diff(rho, p)): sp.diff(rho, p),
    sp.conjugate(sp.diff(C_st, x)): sp.diff(C, x),
    sp.conjugate(sp.diff(C_st, p)): sp.diff(C, p),
    sp.conjugate(sp.diff(sp.diff(C, x), x)): sp.diff(sp.diff(C_st, x), x),
    sp.conjugate(sp.diff(sp.diff(C, p), p)): sp.diff(sp.diff(C_st, p), p),
    sp.conjugate(sp.diff(sp.diff(C, x), p)): sp.diff(sp.diff(C_st, x), p),
    sp.conjugate(sp.diff(sp.diff(C_st, x), x)): sp.diff(sp.diff(C, x), x),
    sp.conjugate(sp.diff(sp.diff(C_st, p), p)): sp.diff(sp.diff(C, p), p),
    sp.conjugate(sp.diff(sp.diff(C_st, x), p)): sp.diff(sp.diff(C, x), p),
    sp.conjugate(sp.diff(sp.diff(rho, x), x)): sp.diff(sp.diff(rho, x), x),
    sp.conjugate(sp.diff(sp.diff(rho, p), p)): sp.diff(sp.diff(rho, p), p),
    sp.conjugate(sp.diff(sp.diff(rho, x), p)): sp.diff(sp.diff(rho, x), p),
}

for replacement in replacement_dict.items():
    rho_0_j_re_conj = rho_0_j_re_conj.replace(*replacement)
```
