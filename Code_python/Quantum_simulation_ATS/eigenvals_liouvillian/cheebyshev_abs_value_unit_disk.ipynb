```python
import numpy as np
from numpy import ma
from scipy.special import chebyt

import matplotlib.pyplot as plt

def plot_abs_chebyshev_on_unit_disk(n, res=400, cmap='viridis'):
    """
    Plot |T_n(z)| for z in the unit disk (Re,Im in [-1,1]).
    """
    xs = np.linspace(-1, 1, res)
    ys = np.linspace(-1, 1, res)
    X, Y = np.meshgrid(xs, ys)
    Z = X + 1j * Y

    inside = X**2 + Y**2 <= 1
    T = chebyt(n)(Z)             # evaluates Chebyshev polynomial on array
    A = np.abs(T)

    A_masked = ma.array(A, mask=~inside)

    plt.figure(figsize=(6, 6))
    pcm = plt.pcolormesh(xs, ys, A_masked, shading='auto', cmap=cmap)
    plt.gca().set_aspect('equal')
    plt.colorbar(pcm, label=f'|T_{n}(z)|')
    plt.gca().add_patch(plt.Circle((0, 0), 1, edgecolor='k', facecolor='none', lw=1))
    cs = plt.contour(X, Y, A_masked, levels=[1], colors='r', linewidths=1.5)
    plt.clabel(cs, fmt=f'|T_{n}|=1', fontsize=8)
    plt.xlabel('Re(z)')
    plt.ylabel('Im(z)')
    plt.title(f'Absolute value of Chebyshev polynomial T_{n} on the unit disk')
    plt.show()

# Example usage:
# plot_abs_chebyshev_on_unit_disk(8)
```
