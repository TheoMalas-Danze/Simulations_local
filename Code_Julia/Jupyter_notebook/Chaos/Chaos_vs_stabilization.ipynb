```julia
const ∞ = Inf
ħ=1
e=1
β = ∞ # Case zero temperature T=0, β → ∞
ϕ_quantum=1/2

#### Caracterization of the circuit #####

ω_a = 3.7*2*pi
ω_b_0 = 7.05*2*pi

Lamb_shift = - 0.05 #Lamb shift in GHz

ω_b = ω_b_0 #+ Lamb_shift*2*pi # Shifted frequency of mode b


φ_a = 0.11
Z_a = 2*ϕ_quantum^2*φ_a^2
C_a_inv = Z_a*ω_a
E_Ca = C_a_inv/2

φ_b = 0.2
Z_b = 2*ϕ_quantum^2*φ_b^2
C_b_inv = Z_b*ω_b
E_Cb = C_b_inv/2

E_J = 37*2*pi




κ_b=1/10*2pi; # 1/ns from Leon's paper
```

```julia
t_max=20/κ_b
```

### Non chaotic 

```julia
ϵ_0 = 0.6
g2_max = besselj(1,ϵ_0)*E_J*φ_a^2*φ_b # (Leon's paper)
ϵ_d=8*g2_max

κ_2 = 4*g2_max^2/κ_b
κ_a = 3e-3*κ_2; #from Leon's paper
```

```julia
ϕ_a, _, q_a, _, t_list = two_photon_CL(t_max;
        circuit_params=(ω_a=ω_a,ω_b=ω_b_0,E_J = E_J, E_Ca=E_Ca, E_Cb=E_Cb, ϵ_0 = ϵ_0,κ_a=0,κ_b=κ_b,ϵ_d=ϵ_d,ω_d=ω_b_0),
        init_mode_a=(ϕ_a_0=0.0,q_a_0 = 0.0), init_mode_b=(ϕ_b_0=0.0,q_b_0 = 0.0), β=∞)

ϕ_rotating_a =@. ϕ_a * cos(ω_a * t_list) + q_a*Z_a* sin(-ω_a * t_list)
q_rotating_a =@. -ϕ_a/Z_a * sin(-ω_a * t_list) + q_a* cos(ω_a * t_list)

n = (q_rotating_a.^2*Z_a + ϕ_rotating_a.^2/Z_a)/2;
```

### Chaotic 

```julia
ϵ_0 = 0.5*pi
g2_max = besselj(1,ϵ_0)*E_J*φ_a^2*φ_b # (Leon's paper)
ϵ_d=10*g2_max

κ_2_chaos = 4*g2_max^2/κ_b
κ_a = 3e-3*κ_2_chaos; #from Leon's paper
```

```julia
ϕ_chaos_a, _, q_chaos_a, _, t_list_chaos = two_photon_CL(t_max;
        circuit_params=(ω_a=ω_a,ω_b=ω_b_0,E_J = E_J, E_Ca=E_Ca, E_Cb=E_Cb, ϵ_0 = ϵ_0,κ_a=0,κ_b=κ_b,ϵ_d=ϵ_d,ω_d=ω_b_0),
        init_mode_a=(ϕ_a_0=0.0,q_a_0 = 0.0), init_mode_b=(ϕ_b_0=0.0,q_b_0 = 0.0), β=∞)

ϕ_rotating_a_chaos =@. ϕ_chaos_a * cos(ω_a * t_list_chaos) + q_chaos_a*Z_a* sin(-ω_a * t_list_chaos)
q_rotating_a_chaos =@. -ϕ_chaos_a/Z_a * sin(-ω_a * t_list_chaos) + q_chaos_a* cos(ω_a * t_list_chaos)

n_chaos = (q_rotating_a_chaos.^2*Z_a + ϕ_rotating_a_chaos.^2/Z_a)/2;
```

```julia
plot(t_list_chaos*κ_2_chaos,n_chaos, label = "Chaotic behavior", color=Perso_palette_hex[3]) 
#plot!(t_list*κ_2,n, label = "Cat stabilized")
plot!(ylabel=L"n", xlabel=L"t/\kappa_2", xlims=(0, 10), ylims=(0,.5))
```

```julia
plot(t_list_chaos*κ_b,n_chaos, label = "Chaotic behavior", color=Perso_palette_hex[3]) 
plot!(t_list*κ_b,n, label = "Cat stabilized")

plot!(ylabel=L"n", xlabel=L"t/\kappa_b")
```
