```julia
using DifferentialEquations, LinearAlgebra, Plots, DiffEqPhysics, StaticArrays


function hamiltonian_osc_verlet(p, q, params, t)
    Mq = params.Mq
    Mq_inv= inv(Mq)
    Mϕ = params.Mϕ
    return 0.5 * (p' * Mq_inv * p + q' * Mϕ * q)
end
```

```julia


function chain_solver(ω,g,C,n,tspan, ϕ_init, q_init, dt=1/(10*maximum(ω)))
    """
    ω : vector of onsite energies
    g : vector of hopping energies
    C : vector of capacitances
    n : number of sites
    tspan = (tmin, tmax) : time span
    ϕ_init : initial value for the position of system oscillator
    q_init : initial value for the momentum of the system oscillator
    dt : time step for the solver, default is 1/(10*maximum(ω)) to verify Shannon's theorem
    """

    Mq = Diagonal(1 ./C)
    Mϕ = SymTridiagonal(C.*(ω.^2), g)


    params = (Mq = Mq, Mϕ = Mϕ)

    ϕ_0 = vcat(SVector(ϕ_init),SVector{n}(zeros(n)))
    q_0 = vcat(SVector(q_init),SVector{n}(zeros(n)))

    a = HamiltonianProblem(hamiltonian_osc, q_0, ϕ_0, tspan, params)

    sol = solve(a, VelocityVerlet(),dt = dt)

    return sol,dt
end
```

```julia
function compute_total_energy(sol, ω,g,C, n)
    """
    sol : solution of the Hamiltonian system
    params : parameters of the Hamiltonian system
    """

    Mq = Diagonal(1 ./C)
    Mϕ = SymTridiagonal(C.*(ω.^2), g)

    q = [getindex.(sol.u,i) for i in 1:n+1]
    ϕ = [getindex.(sol.u,i+n+1) for i in 1:n+1]

    E = 0.5 * [getindex.(q,t)' * Mq * getindex.(q,t) + getindex.(ϕ,t)' * Mϕ * getindex.(ϕ,t) for t in 1:length(sol.t)];

    return E
end

function compute_total_energy(X::Vector, ω,g,C, n)
    """
    X : vector of dynamics of the Hamiltonian system
    params : parameters of the Hamiltonian system
    """

    Mq = Diagonal(1 ./C)
    Mϕ = SymTridiagonal(C.*(ω.^2), g)

    ϕ = [getindex.(X,i) for i in 1:n+1]
    q = [getindex.(X,i+n+1) for i in 1:n+1]

    E = 0.5 * [getindex.(q,t)' * Mq * getindex.(q,t) + getindex.(ϕ,t)' * Mϕ * getindex.(ϕ,t) for t in 1:length(X)];

    return E
end
```

```julia
function solve_normal_modes(ω, g, C, ϕ_0, q_0, tspan; dt=1/(10*maximum(ω)), energy=false)
    """ 
    #INPUT
    ω : vector of onsite energies
    g : vector of hopping energies
    C : vector of capacitances
    ϕ_0 : initial value for the position of system oscillator
    q_0 : initial value for the momentum of the system oscillator
    tspan = (tmin, tmax) : time span
    dt : time step for the solver, default is 1/(10*maximum(ω)) to verify Shannon's theorem
    energy : boolean, if true compute the energy of the system

    #OUTPUT
    tlist : time list
    ϕ_t : time evolution of the position of system oscillator
    q_t : time evolution of the momentum of system oscillator
    E : time evolution of the energy of the system if energy=true
    """
    npoints = Int(round((tspan[2]-tspan[1])/dt))  # number of time points

    # Build M1 and M2
    M1 = Diagonal(1.0 ./ C)
    inv_M1 = Diagonal(C)
    d = C .* ω.^2
    M2 = SymTridiagonal(-d, -g)

    # A = M1 * M2, to be diagonalized
    A = M1 * M2
    vals, vecs = eigen(A)

    # Frequencies and mode vectors
    ωk² = -real(vals)

    if count( x -> (x < 0), ωk²) > 0
        display(print(vecs*ϕ_0))
        display(scatter(ωk², label="Eigenvalues"))
        error("Negative eigenvalues found in the system.")
    end

    ωk = sqrt.(ωk²)
    Φk = vecs  # columns = normal modes
    Φk_inv=inv(Φk)
    # Project initial conditions into normal mode basis
    a0 = Φk_inv * ϕ_0
    adot0 = Φk_inv * M1 * q_0

    # Time evolution in normal coordinates
    tlist = range(tspan[1], tspan[2], length=npoints)
    a_t = [a0 .* cos.(ωk .* t) .+ (adot0 ./ ωk) .* sin.(ωk .* t) for t in tlist]
    a_t_dot= [a0 .* (-ωk) .* sin.(ωk .* t) .+ (adot0) .* cos.(ωk .* t) for t in tlist]

    # Reconstruct real-space fields
    ϕ_t = [Φk * a for a in a_t]
    q_t = [inv_M1 * Φk * a for a in a_t_dot]  # q = \dot{ϕ}/C


    if energy
        # Compute energy
        E = 0.5 * [q_t[t]' * M1 * q_t[t] + ϕ_t[t]' * (-M2) * ϕ_t[t] for t in 1:npoints]
        return tlist, ϕ_t, q_t,E
    end

    return tlist, ϕ_t, q_t
end
```

## Modelisation of the Caldeira-Leggett model

We rely on the article "Thorbeck, Ted, Alexander McDonald, O. Lanes, John Blair, George Keefe, Adam A. Stabile, Baptiste Royer, Luke C. G. Govia, et Alexandre Blais. « High-Fidelity Gates in a Transmon Using Bath Engineering for Passive Leakage Reset ». arXiv, 6 novembre 2024. https://doi.org/10.48550/arXiv.2411.04101." The bath is colored via the purcell filter, creating a pass-band filter. The frequency absorption for the bath is between $\omega_{min}=$ 2.5 GHz and $\omega_{max}=$ 4.5 GHz. We will take $\Re(Z)=Z_0$ in the window and 0 outside.

For a Caldeira-Leggett approach: $\omega_m=\omega_{min}+m\frac{\omega_{max}-\omega_{min}}{n}$, $C_m=\frac{\Re(Z)(\omega_m)}{\omega_m}=Z_0/\omega_m$.

For the oscillator, we set the frequency in the middle of the window $\omega_0=3.5$ GHz and we choose the same capacitance order of magnitude as the one given for the modes $a$ and $b$ in Lescanne, Raphaël, Marius Villiers, Théau Peronnin, Alain Sarlette, Matthieu Delbecq, Benjamin Huard, Takis Kontos, Mazyar Mirrahimi, et Zaki Leghtas. « Exponential Suppression of Bit-Flips in a Qubit Encoded in an Oscillator ». Nature Physics 16, nᵒ 5 (mai 2020): 509‑13. https://doi.org/10.1038/s41567-020-0824-x, so $E_C=100$ MHz.

It will be easier to work with $1/C=2E_C/e^2$ and $1/C_m=\omega_m/Z_0$.

The capacitance energy given in Lescanne, Raphaël, Marius Villiers, Théau Peronnin, Alain Sarlette, Matthieu Delbecq, Benjamin Huard, Takis Kontos, Mazyar Mirrahimi, et Zaki Leghtas. « Exponential Suppression of Bit-Flips in a Qubit Encoded in an Oscillator ». ## Finding the normal modes and the coupling through the spectral density function. We map the Caldeira-Leggett model onto a 1d semi-infinite chain

```julia
using MPSDynamics
```

```julia
function smoothstep(x)
    return x^3 * (x * (x * 6 - 15) + 10)
end
```

```julia
const ∞ = Inf
ħ=1
e=1
β = ∞ # Case zero temperature T=0, β → ∞
n = 100 # length of the chain

ω_max = 10.0 *2*pi # GHz
ω_min = 2.0 *2*pi # GHz

#### Caracterization of the LC circuit #####
ω_sys = (ω_max+ω_min)/2 # GHz
C_sys_inv = 9 #ħ/e^2 GHz from Lescanne et al.
L_sys=C_sys_inv/ω_sys^2 #ħ/e^2 /GHz



### Caracterization of the impedance #####
κ=1/20 # ns from Jeffrey et al. 
Z = C_sys_inv/(2*κ) #ħ/e^2

# We design the impedance to be a smooth pass-band filter
ωc_dwn = ω_min-.2*2*pi # Lower cut-off of the spectral density J(ω), GHz
ωc_up = ω_max+.2*2*pi # Cut-off of the spectral density J(ω), GHz

function Imped(ω)
    if ω < ωc_dwn
        return 0.0
    elseif ω < ω_min
        return Z * smoothstep((ω - ωc_dwn) / (ω_min - ωc_dwn))
    elseif ω< ω_max
        return Z
    elseif ω < ωc_up
        return Z * (1 - smoothstep((ω - ω_max) / (ωc_up - ω_max)))
    else
        return 0.0
    end
end


```

```julia
Z_list=0:.1:100
plot(Z_list, [Imped(i) for i in Z_list], label="Z(ω)", xlabel="ω")
```

```julia
# AB segments correspond to the different frequency ranges where the spectral density is defined.
# It corresponds to i=1 ; i=2 and i=3 of the Jω_fct(ω,i) function.

AB = [[-Inf ωc_dwn];[ωc_dwn ω_min];[ω_min ω_max];[ω_max ωc_up];[ωc_up Inf]]
#AB=[[-Inf ω_min];[ω_min ω_max];[ω_max Inf]]


function Jω_fct(ω,i)
    if i==1 # First segment of AB (here [-Inf ω_min]), zero for negative frequencies when β == Inf
        y = 0
    elseif i==2 # Second segment of AB (here [ω_min ω_max])
        y = pi*Imped.(ω).*ω.^2
    elseif i==3 # Third segment of AB (here [ω_min ω_max])
        y = pi*Imped.(ω).*ω.^2
    elseif i==4 # Fourth segment of AB (here [ω_max ωc_up])``
        y = pi*Imped.(ω).*ω.^2
    elseif i==5 # Fifth segment of AB (here [ωc_up Inf])
        y = 0
    end
    return y
end
```

```julia
onsite_energies,hopping_energies,system_chain_coupling = chaincoeffs_finiteT(n, β, false; J=Jω_fct, mc=size(AB)[1], mp=0, AB=AB, iq=1, idelta=2, procedure=:Lanczos, Mmax=5000, save=false)  # chain parameters, i.e. on-site energies ϵ_i, hopping energies t_i, and system-chain coupling c_0
```

```julia
p1=scatter(onsite_energies)
p2=scatter(hopping_energies)
plot(p1,p2,layout=(1,2),size=(1200,400),legend=:topright)
```

```julia
### Caldeira Leggett model ###

C_sys_prime_inv = C_sys_inv + Z*(ωc_up +ω_max - ω_min - ωc_dwn)/2 #ħ/e^2 GHz
ω_sys_prime = sqrt((C_sys_prime_inv/L_sys)) # GHz
Z_sys_prime = sqrt(L_sys*C_sys_prime_inv)

### Mapping ###
ω = vcat(ω_sys_prime,onsite_energies) # GHz
g_q=vcat(-system_chain_coupling/sqrt(Z_sys_prime), -hopping_energies) 
g_ϕ=vcat(0.0, -hopping_energies) 

#C=vcat(C_sys_prime,1 ./(Z*onsite_energies)) #ħ/e^2 GHz

ϕ_init = 1.0
q_init = 0.0

ϕ_0 = vcat(SVector(ϕ_init),SVector{n}(zeros(n)))
q_0 = vcat(SVector(q_init),SVector{n}(zeros(n)))

tspan = (0.0, 20π / minimum(ω))

Mq=SymTridiagonal(ω, g_q)
Mϕ=SymTridiagonal(ω, g_ϕ)

A=Mq * Mϕ
vals, vecs = eigen(A)

#ωk² = (vals)
#inv_vecs = inv(vecs)
#scatter(ωk²[1:end], label="Eigenvalues")



#t, ϕ_t, q_t,E = solve_normal_modes(ω, g, C, ϕ_0, q_0, tspan, energy=true)
#p1=plot(sol; legend=:outertopright)
#p2=plot(compute_total_energy(sol,ω, g, C, N))
#plot(p1,p2,layout=(1,2), size=(1200,400))
```

```julia
function first_row_matrix(v::Vector)
    n = length(v)
    M = zeros(eltype(v), n, n)
    M[1, :] .= v
    return M
end
function first_column_matrix(v::Vector)
    n = length(v)
    M = zeros(eltype(v), n, n)
    M[:, 1] .= v
    return M
end
```

```julia
const ∞ = Inf
ħ=1
e=1
β = ∞ # Case zero temperature T=0, β → ∞
n = 100 # number of connected sites

ω_max = 4.5 *2*pi # GHz
ω_min = 2.5 *2*pi # GHz

#### Caracterization of the LC circuit #####
ω_sys = (ω_max+ω_min)/2 # GHz
C_sys_inv = 0.147*2*pi #ħ/e^2 GHz from Lescanne et al.
L_sys=C_sys_inv/ω_sys^2 #ħ/e^2 /GHz



### Caracterization of the impedance #####
κ=1/20 # GHz from Jeffrey et al. 
Z = C_sys_inv/(2*κ) #ħ/e^2

### Caldeira Leggett model ###
ω_m=ω_min:(ω_max-ω_min)/(n-1):ω_max
C_m_inv=Z*ω_m

C_sys_prime = 1/(C_sys_inv + sum(C_m_inv)/2) #ħ/e^2 GHz
ω_sys_prime = sqrt(1/(C_sys_prime*L_sys)) # GHz


Mq=Diagonal(vcat(1.0 ./ C_sys_prime,1.0 ./ C_m_inv))+first_row_matrix(vcat(0.0,-1.0 ./ C_m_inv))+first_column_matrix(vcat(0.0,-1.0 ./ C_m_inv))
Mϕ=Diagonal(vcat(ω_sys^2/C_sys_inv,ω_m.^2 ./C_m_inv))

inv_Mq=inv(Mq);
```

```julia
A=Mq * Mϕ
vals, vecs = eigen(A)

ωk² = real(vals)
inv_vecs = inv(vecs)
scatter(ωk²[1:end], label="Eigenvalues")
```
